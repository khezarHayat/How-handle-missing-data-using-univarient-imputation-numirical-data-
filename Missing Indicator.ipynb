```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv('Titanic.csv',usecols=['Age','Fare','Survived'])
```

```python
df
```

```python
from sklearn.model_selection import train_test_split
from sklearn.impute import MissingIndicator,SimpleImputer
```

```python
x=df.drop(columns=['Survived'])
y=df['Survived']
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
x_train
```

```python
# using simple imputer  we will compare it with missing indicator
si=SimpleImputer()
x_train_trf=si.fit_transform(x_train)
x_test_trf=si.transform(x_test)
```

```python
x_train_trf
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score 
clf=LogisticRegression()
clf.fit(x_train_trf,y_train)
y_pred=clf.predict(x_test_trf)
```

```python
accuracy_score(y_test,y_pred)
```

```python
# now we using missing indicator
mi=MissingIndicator()
mi.fit(x_train)
```

```python
mi.features_        # this tell wear is the missing value show in 1 col the missing value s present 
```

```python
x_train_missing=mi.transform(x_train)
```

```python
x_train_missing    #it is the new col which have new vlues t,f
```

```python
x_test_missing=mi.transform(x_test)
```

```python
# now we make new col and add all this new values
x_train['Age_miss']=x_train_missing
x_test['Age_miss']=x_test_missing
```

```python
x_train           # now see the indicator col which have if value is nul show true if present show false
```

```python
# now if the missing indicator is usd
si=SimpleImputer()

x_train_trf2=si.fit_transform(x_train)
x_test_trf2=si.transform(x_test)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score 
clf=LogisticRegression()
clf.fit(x_train_trf2,y_train)
y_pred=clf.predict(x_test_trf2)
```

```python
accuracy_score(y_test,y_pred)
```

```python
# we also used direct parameter of idicator with out using the missing indicator class separate
```
